## 1. Import required libraries:

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

## 2. Load Customer and Transaction Data:

```python
customer_datasets = pd.read_csv("customer_datasets.csv")
customer_datasets
```

```python
transaction_datasets = pd.read_csv("datasets.csv")
transaction_datasets
```

1. Clean column names

```python
customer_datasets .columns = customer_datasets.columns.str.strip()
transaction_datasets.columns = transaction_datasets.columns.str.strip()
```

2. Check the first few rows:

```python
customer_datasets.head()
transaction_datasets.head()
```

 # 3.  Merge Both Datasets:

```python
Merge_df= pd.merge(transaction_datasets, customer_datasets, how='left', left_on='Credit_card', right_on='C_ID')
Merge_df
```

# 4. Calculate the spend in terms of Product, State and Payment method.     

1. Calculate Spending:

```python
Merge_df['Spend'] = Merge_df['Selling_price']
Merge_df
```

2. Group by Product, State, and Payment Method

```python
spend_summary = Merge_df.groupby(['Product_ID', 'State', 'Payment Method'])['Spend'].sum().reset_index()
spend_summary
```

3. Highest 5 spenders in each group

```python
highest_5 = spend_summary.sort_values(by='Spend', ascending=False).head(5)
highest_5
```

# 5. Return Category Analysis:

1. Return datasets.

```python
return_data = Merge_df[Merge_df['Return_ind'] == 1]
return_data
```

2. Return pattern by state and age

```python
return_data.groupby('State')['Return_ind'].count().sort_values(ascending=False)
return_data.groupby('Age')['Return_ind'].count().sort_values(ascending=False)
```

3. If Selling_price is far below Price, it means a big discount — check correlation.

```python
# Convert columns to numeric (ignore any non-numeric values)
Merge_df['Price'] = pd.to_numeric(Merge_df['Price'], errors='coerce')
Merge_df['Selling_price'] = pd.to_numeric(Merge_df['Selling_price'], errors='coerce')

# Calculate Discount
Merge_df['Discount'] = Merge_df['Price'] - Merge_df['Selling_price']

# Check correlation between Return_ind and Discount
print(Merge_df[['Return_ind', 'Discount']].corr())
```

# 6. Create a profile of customers in terms of timing of their order:

```python
Merge_df['Hour'] = pd.to_datetime(Merge_df['Time'], format='%H:%M:%S').dt.hour
hourly_orders = Merge_df.groupby('Hour')['Transaction ID'].count()
plt.plot(hourly_orders.index, hourly_orders.values)
plt.title('Orders per Hour of Day', fontsize=25,color="pink")
plt.xlabel('Hour',fontsize=10,color="red")
plt.ylabel('Number of Orders',fontsize=10,color="red")
plt.show()
```

# 7. Which Payment Method Offers More Discount:

```python
payment_discount = Merge_df.groupby('Payment Method')['Discount'].mean().sort_values(ascending=False)
payment_discount
```

# 8. High vs. Low Value Items Profile

1. Average Price:

```python
avg_price = Merge_df['Selling_price'].mean()
Merge_df['Value_Category'] = Merge_df['Selling_price'].apply(lambda x: 'High' if x > avg_price else 'Low')
```

2. Orders by category:

```python
Merge_df['Value_Category'].value_counts()
```

3. Compare discount and count per category: 

```python
Merge_df.groupby('Value_Category')[['Discount', 'Transaction ID']].agg({'Discount': 'mean', 'Transaction ID': 'count'})
```

# 9. Do Discounts Increase Orders 

```python
discount_order = Merge_df.groupby('Discount')['Transaction ID'].count().reset_index()
plt.scatter(discount_order['Discount'], discount_order['Transaction ID'])
plt.title('Effect of Discount on Order Volume',fontsize=25,color="pink")
plt.xlabel('Discount Amount',fontsize=10,color="red")
plt.ylabel('Number of Orders',fontsize=10,color="red")
plt.show()
```
